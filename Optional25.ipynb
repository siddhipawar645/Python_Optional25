```python
import pandas as pd
import numpy as np
import joblib
from sklearn.model_selection import train_test_split, GridSearchCV, RandomizedSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR
from sklearn.ensemble import (
    RandomForestRegressor,
    AdaBoostRegressor,
    GradientBoostingRegressor
)
from xgboost import XGBRegressor
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)
```

```python
df = pd.read_csv("Gold_Dataset.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.fillna(df.mean(numeric_only=True), inplace=True)

df.isnull().sum()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for col in df.columns:
    if df[col].dtype == "object":
        df[col] = le.fit_transform(df[col])

df.head()
```

```python
X = df.drop("GLD", axis=1)
y = df["GLD"]
```

```python
scaler = StandardScaler()

X = scaler.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.33,
    random_state=42
)
```

Q1


```python
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import r2_score
```

```python
k_values = [3, 5, 7, 11, 13, 15]

best_score = -1
best_k = None
```

```python
for k in k_values:

    knn = KNeighborsRegressor(n_neighbors=k)

    knn.fit(X_train, y_train)

    y_pred = knn.predict(X_test)

    score = r2_score(y_test, y_pred)

    print(f"K = {k}, R2 Score = {score:.4f}")

    if score > best_score:
        best_score = score
        best_k = k
```

```python
print("Best K Value :", best_k)
print("Best R2 Score :", round(best_score, 4))
```

Q2

```python
from sklearn.svm import SVR
from sklearn.metrics import r2_score
```

```python
C_values = [1, 10, 20]
kernels = ["linear", "rbf"]

best_score = -1
best_C = None
best_kernel = None
```

```python
for C in C_values:
    for kernel in kernels:

        svr = SVR(C=C, kernel=kernel)

        svr.fit(X_train, y_train)

        y_pred = svr.predict(X_test)

        score = r2_score(y_test, y_pred)

        print(f"C = {C}, Kernel = {kernel}, R2 Score = {score:.4f}")

        if score > best_score:
            best_score = score
            best_C = C
            best_kernel = kernel
```

```python
print("Best C Value :", best_C)
print("Best Kernel :", best_kernel)
print("Best R2 Score :", round(best_score, 4))
```

Q3

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    "C": [1, 10, 20],
    "kernel": ["linear", "rbf"]
}
```

```python
grid = GridSearchCV(
    estimator=SVR(),
    param_grid=param_grid,
    cv=5,
    scoring="r2"
)
```

```python
grid.fit(X_train, y_train)
```

```python
print("Best Parameters :", grid.best_params_)
print("Best Cross Validation Score :", round(grid.best_score_, 4))
```

```python
best_svr = grid.best_estimator_

y_pred = best_svr.predict(X_test)

r2 = r2_score(y_test, y_pred)

print("Test R2 Score :", round(r2, 4))
```
